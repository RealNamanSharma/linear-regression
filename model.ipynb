```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# x_train: Hours studied per week
x_train = np.array([1.5, 2.0, 2.8, 3.5, 4.0, 4.8, 5.2, 6.0, 6.7, 7.5, 8.2, 9.0, 9.8, 10.5, 11.2])

# y_train: Exam score (out of 100)
y_train = np.array([38.2, 42.1, 47.5, 51.0, 56.4, 60.8, 63.5, 69.2, 72.0, 78.4, 81.1, 86.5, 91.0, 93.8, 98.2])
```

```python
plt.scatter(x_train, y_train)
plt.xlabel('hours studied per week')
plt.ylabel('exam score')
```

```python
def compute_cost(x, y, w, b):
    m = np.shape(x)[0]
    cost = 0
    for i in range(m):
        val = ((w * x[i] + b - y[i]) ** 2)
        cost += val
    cost = cost / 2 * m
    return cost
```

```python
def compute_grad(x, y, w, b):
    m = np.shape(x)[0]
    df_dw = 0
    df_db = 0
    for i in range(m):
        df_dw += (w * x[i] + b - y[i]) * y[i]
        df_db += (w * x[i] + b - y[i])
    df_dw /= m
    df_db /= b
    return df_dw, df_db
```

```python
def gradient_descent(x, y, w_init, b_init, alpha, itr, compute_grad, compute_cost):
    w_final = 0
    b_final = 0
    w = w_init
    b = b_init
    m = np.shape(x)[0]
    df_dw, df_db = compute_grad(x, y, w_init, b_init)
    for i in range(itr):
        w = w - alpha * df_dw
        b = b - alpha * df_db
        df_dw, df_db = compute_grad(x, y, w, b)
        if i % 10000 == 0:
            print(f"for w = {w} and b = {b} cost = ",compute_cost(x, y, w, b))
    w_final =  w
    b_final = b
    return w_final, b_final
```

```python
w, b = gradient_descent(x_train, y_train,10,10, 0.001, 90000, compute_grad, compute_cost)
print(f"w = {w:.3f}, b = {b:.3f} are optimal values")
```

```python
plt.scatter(x_train, y_train, color="blue")
y_pred = w * x_train + b
plt.plot(x_train, y_pred, color="red")
plt.xlabel('hours studied per week')
plt.ylabel('exam score')
```
